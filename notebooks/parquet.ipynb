```python
from pathlib import Path
import duckdb
```

```python

path = Path("/home/sadhika8/JupyterLinks/nobackup/quads_results/GEOSFP/2024/03/01/quads_results_geosfp_2024_03_01_violations_raw_data.parquet").resolve()

con = duckdb.connect()

print("\n--- File ---")
print(path)

print("\n--- Schema ---")
print(con.execute(
    "DESCRIBE SELECT * FROM read_parquet(?)",
    [str(path)]
).df())

print("\n--- First 10 rows ---")
print(con.execute(
    "SELECT * FROM read_parquet(?) LIMIT 10",
    [str(path)]
).df())

print("\n--- Row count ---")
print(con.execute(
    "SELECT COUNT(*) AS n_rows FROM read_parquet(?)",
    [str(path)]
).df())

print("\n--- Summary ---")
print(con.execute(
    "SUMMARIZE SELECT * FROM read_parquet(?)",
    [str(path)]
).df())
```

```python
target_id = "inst3_3d_chm_Nv|CO2|1.0|Strat2"

df = con.execute("""
SELECT *
FROM read_parquet(?)
WHERE id_string = ?
LIMIT 100
""", [str(path), target_id]).df()

print(df)
```

```python
print(con.execute("""
SELECT time, COUNT(*) AS n
FROM read_parquet(?)
WHERE id_string = ?
GROUP BY time
ORDER BY time
""", [str(path), target_id]).df())
```
